```python
#  que -8)Implement program to count Uppercase, Lowercase, special characters and numeric values 
# in a given string. 
```

```python
a="Hello@1234";
Upcnt=0;
locnt=0;
spcharcnt=0;
numcnt=0
for i in range(0,len(a)):
    if a[i].isupper():
        Upcnt=Upcnt+1;
    elif a[i].islower():
        locnt=locnt+1;
    elif a[i].isdigit():
        numcnt=numcnt+1;
    else:
        spcharcnt=spcharcnt+1;
print("Uppercnt",Upcnt)
print("lowerCnt",locnt)
print("specialchar",spcharcnt)
print("numericcnt",numcnt)
        
       
        
```

```python
# Que 16).. Write a Python function that takes a list of words and return the longest word and the length 
# of the longest one 
```

```python
n=input("Enetr the words");
lst=list(n.split());
def maxim(lst,maxi,maxwor):
    for ch in lst:
        if len(ch)>maxi:
            maxi=len(ch);
            maxwor=ch;
    return [maxi,maxwor];
print(maxim(lst,0,""));
```

```python
 # Que 20) Write a Python function to reverse a string if its length is a multiple of 4. 
```

```python
st="helloooo"
def rev(st):
    if len(st)%4==0:
        return st[::-1]
    else:
        return st;

print(rev(st));       
       
```
